#### <font color = 'teal'>K nearest neighbor (K-NN)

#### <font color = 'teal'> Fonctionnement mathématique du K-NN

Le **K plus proches voisins (K-Nearest Neighbors)** est un algorithme d'apprentissage supervisé **non paramétrique** et **paresseux** (*lazy learning*) : il ne construit aucune fonction/modèle explicite pendant l'entraînement, il se contente de **mémoriser** l'ensemble des données d'entraînement $(x_i, y_i)$. Tout le calcul se fait au moment de la prédiction.

**1. Principe général**

Pour prédire la classe d'une nouvelle observation $x$ :
1. On calcule la distance entre $x$ et **tous** les points $x_i$ de l'ensemble d'entraînement.
2. On sélectionne les $k$ points les plus proches de $x$ (les "$k$ voisins").
3. On attribue à $x$ la classe **majoritaire** parmi ces $k$ voisins (vote majoritaire).

**2. Mesure de distance**

La distance la plus utilisée (et celle utilisée par défaut dans `KNeighborsClassifier`) est la **distance euclidienne**. Pour deux observations $x = (x^{(1)}, x^{(2)}, ..., x^{(p)})$ et $x_i = (x_i^{(1)}, x_i^{(2)}, ..., x_i^{(p)})$ décrites par $p$ variables :

$$d(x, x_i) = \sqrt{\sum_{j=1}^{p} \left(x^{(j)} - x_i^{(j)}\right)^2}$$

D'autres distances existent (paramètre `p` de Minkowski) :
- **Manhattan** ($p=1$) : $d(x, x_i) = \sum_{j=1}^{p} \left| x^{(j)} - x_i^{(j)} \right|$
- **Minkowski (générale)** : $d(x, x_i) = \left( \sum_{j=1}^{p} \left| x^{(j)} - x_i^{(j)} \right|^q \right)^{1/q}$, l'Euclidienne étant le cas $q=2$.

⚠️ C'est précisément pour que cette distance ait un sens (et que les variables avec de grandes échelles, comme `Prix`, ne dominent pas les autres) qu'on a **normalisé** les variables avec `MinMaxScaler` avant d'entraîner le modèle.

**3. Règle de décision (classification)**

Soit $N_k(x)$ l'ensemble des $k$ voisins les plus proches de $x$. La classe prédite $\hat{y}$ est celle qui maximise la probabilité *a posteriori* estimée par le vote majoritaire :

$$\hat{y} = \underset{c}{\text{argmax}} \sum_{i \in N_k(x)} \mathbb{1}(y_i = c)$$

où $\mathbb{1}(y_i = c)$ vaut $1$ si le voisin $i$ appartient à la classe $c$, et $0$ sinon. Dans notre cas binaire ($c \in \{0 : \text{D'occasion},\ 1 : \text{Venant}\}$), on compte simplement combien de voisins sont dans chaque classe et on choisit la classe majoritaire.

On peut aussi utiliser un **vote pondéré** (`weights = 'distance'`), où chaque voisin pèse d'autant plus qu'il est proche :

$$w_i = \frac{1}{d(x, x_i)}$$

**4. Choix du nombre de voisins $k$**

$k$ est un **hyperparamètre** clé :
- $k$ trop petit → modèle très sensible au bruit (overfitting), frontière de décision très irrégulière.
- $k$ trop grand → modèle trop lisse, risque de sous-apprentissage (underfitting), et les classes minoritaires sont noyées.

C'est pourquoi on utilise ci-dessous une **validation croisée** (`GridSearchCV`, $cv=5$) pour tester $k \in \{2, ..., 30\}$ et choisir la valeur qui maximise l'accuracy moyenne sur les 5 plis.

**5. Complexité**

Comme il n'y a pas d'entraînement à proprement parler, la prédiction pour une seule observation coûte $O(n \times p)$ (calcul de $n$ distances sur $p$ variables), $n$ étant le nombre d'observations d'entraînement : c'est pour cela que le K-NN peut devenir lent sur de grands jeux de données.

```python
# Importer KNeighborsClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
# importer GridSearchCV
from sklearn.model_selection import GridSearchCV
# initialiser le modèle knn
knn = KNeighborsClassifier()
# choisir une rangée de valeurs variante de 2 à 30
k_list = list(range(2,31))
# créer le dictionnaire params
params = {'n_neighbors' : k_list}
# Instantier le modèle GridSearchCV
grid = GridSearchCV(knn, params, cv = 5)

# lancer la recherche
grid.fit(x_train, y_train)
```

```python
# k optimal
k = grid.best_params_['n_neighbors']
# Instantier le modèle
knn_modele = KNeighborsClassifier(n_neighbors = k)
# Entrainer le modèle
knn_modele.fit(x_train, y_train)
```

### <font color = 'bisque'> Prédiction

```python
#un echantilon de 10 observations sur les donnees
x_test_10 = x_test[10:20]
y_test_10 = y_test[10:20]
# pour la prediction
y_predict_10 = knn_modele.predict(x_test_10)
class_names = ['Ocasion', 'venant']

# afficher les predictions
for i in range(10):
  print(f'prediction {str(i+1)} : {class_names[y_predict_10[i]]} | Reel {str(i+1)} {class_names[y_test_10[i]]}')
```

### <font color = 'bisque'> Evaluation

```python
# Importer accuracy_score
from sklearn.metrics import accuracy_score
y_pred = knn_modele.predict(x_test)
acc_knn = accuracy_score(y_test, y_pred)
# affichage de l'accuracy
print(f'Accuracy du models : {acc_knn: .2f}')

```
